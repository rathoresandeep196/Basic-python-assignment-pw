1. What is Python, and why is it popular ?
- Python is a high-level, interpreted programming language with simple and readable syntax.
It is widely used in the industry because it is easy to learn, supports a large number of libraries, and has a strong developer community.
Python is also cross-platform, which means it works on Windows, macOS, and Linux.


2. What is an interpreter in Python ?
- An interpreter in pyton read and execute code line by line.
it translates the high-level Python code into machine-level code at runtime, allowing you to run the program without compiling it first.it help us to catch the bug early

3. What are pre-defined keywords in Python ?
- Pre-defined keywords in Python are reserved words that have special meaning and functionality defined by the Python interpreter.
We cannot use them as variable names or identifiers in our program.

4. Can keywords be used as variable names ?
- No, we cannot use keywords as variable names in Python because they have special meaning and functionality defined by the language.

5. What is mutability in Python ?
- Mutability in Python means that the value of an object can be changed after it is created.
Examples of mutable data types include: list, dictionary, and set.

6. Why are lists mutable, but tuples are immutable ?
- Lists are mutable because their elements can be changed, added, or removed after the list is created.
Tuples are immutable because once a tuple is created, its elements cannot be changed.This difference exists because lists are designed for dynamic data, while tuples are used for fixed data or read-only collections

8. What are logical operators in Python ?
- Logical operators in Python are used to combine conditional statements.
Python has three logical operators:

- and – returns True if both conditions are true

- or – returns True if at least one condition is true

- not – reverses the result (True becomes False, and vice versa)



9. What is type casting in Python ?
- Type casting in Python means converting a value from one data type to another compatible data type to perform operations.
For example: converting int to float, float to int, string to int, or boolean to int.
Python provides built-in functions for type casting like int(), float(), str(), and bool().

10. What is the difference between implicit and explicit type casting ?
- In Implicit type casting python interpreter automatically convert data types to another data type without developer involvement for example - a=5 , b=5.5 then a+b =5.5 here python will automatically convert 5 into float
- In Explicit type casting we manually type cast the data types for example a=5 , a=float(a) , a=5.0 here we use the built-in function of python to convert int to float

11. What is the purpose of conditional statements in Python ?
- Conditional statements in Python are used to make decisions in a program. They allow the program to execute specific blocks of code only when certain conditions are met, using if, elif, and else statements.

- if: Executes a block of code if the condition is true.

- elif: Executes when the if condition is false and its own condition is true. Multiple elif blocks can be used.

- else: Executes if all if and elif conditions are false.

```python
# 6. Write a function to demonstrate the behavior of mutable and immutable arguments?
def demo_mutable_immutable(mutable_arg, immutable_arg):
    print("Before modification:")
    print("Mutable argument:", mutable_arg)
    print("Immutable argument:", immutable_arg)
    mutable_arg.append(100)
    immutable_arg = 200
    print("\nAfter modification:")
    print("Mutable argument:", mutable_arg)
    print("Immutable argument:", immutable_arg)
my_list = [1, 2, 3]
my_int = 10

demo_mutable_immutable(my_list, my_int)
print("\nOutside function:")
print("Original list:", my_list)
print("Original integer:", my_int)
```

```python
# 14. Write a program to reverse a string using a while loop.?
user_string=input("enter your string ")
reversed_string=""
string_len=len(user_string)
i=0
while string_len > 0:
  reversed_string=user_string[i] + reversed_string
  string_len=string_len-1
  i=i+1
print(reversed_string)
```

```python
# 15. Write a Python program to calculate the factorial of a number provided by the user using a while loop.?
factorial_number=int(input("please enter your number to calculate factorial "))
ans=1
while factorial_number >=1:
  ans=ans*factorial_number
  factorial_number=factorial_number-1
print(ans)

```
